```python
# Q1
def string_ops(s):
    return s[::-1], s[::2]
```

```python
# Q2
def squares_even(lst):
    return [x**2 for x in lst if x % 2 == 0]
```

```python
# Q3
def char_freq(s):
    d = {}
    for c in s:
        d[c] = d.get(c, 0) + 1
    return d
```

```python
# Q4
def odd_indices(lst):
    return [i for i, x in enumerate(lst) if x % 2 != 0]
```

```python
# Q5
def rotate(lst, k):
    return lst[-k:] + lst[:-k]
```

```python
# Q6
def lists_to_dict(a, b):
    return dict(zip(a, b))
```

```python
# Q7
def manual_iter(lst):
    it = iter(lst)
    while True:
        try:
            print(next(it))
        except StopIteration:
            break
```

```python
# Q8
def flatten(lst):
    return [item for sublist in lst for item in sublist]
```

```python
# Q9
def is_palindrome(s):
    return s == s[::-1]
```

```python
# Q10
def second_largest(lst):
    first = second = float('-inf')
    for x in lst:
        if x > first:
            second = first
            first = x
        elif x > second and x != first:
            second = x
    return second
```
